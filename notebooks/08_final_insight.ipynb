```python
# Import libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load processed datasets

weather_df = pd.read_csv("../data/processed/pakistan_weather_2025.csv")

cost_df = pd.read_csv("../data/processed/pakistan_city_cost_of_living_cleaned.csv")

economic_df = pd.read_csv("../data/processed/pakistan_city_economic_summary.csv")

economic_categories_df = pd.read_csv("../data/processed/pakistan_economic_categories.csv")
```

```python
# Check dataset shapes

print("Weather Dataset:", weather_df.shape)
print("Cost of Living Dataset:", cost_df.shape)
print("Economic Summary Dataset:", economic_df.shape)
print("Economic Categories Dataset:", economic_categories_df.shape)
```

```python
# Check missing values in all datasets

print("Weather Missing Values:")
print(weather_df.isnull().sum())

print("\nCost of Living Missing Values:")
print(cost_df.isnull().sum())

print("\nEconomic Summary Missing Values:")
print(economic_df.isnull().sum())

print("\nEconomic Categories Missing Values:")
print(economic_categories_df.isnull().sum())
```

```python
# Overall project statistics

print("PAKINSIGHT - PROJECT OVERVIEW")
print("-" * 40)

print("Total Cities:", weather_df["city"].nunique())

print("Total Weather Records:", len(weather_df))

print("Total Essential Items:", cost_df["Item"].nunique())

print("Total Economic Categories:", economic_categories_df["Description"].nunique())

print("Total Establishments:", economic_df["Establishments"].sum())

print("Total Workforce:", economic_df["Workforce"].sum())
```

```python
# City with highest average maximum temperature
hottest_city = weather_df.groupby("city")["temp_max"].mean().idxmax()

# City with highest total precipitation
rainiest_city = weather_df.groupby("city")["precipitation"].sum().idxmax()

# City with highest establishments
top_establishment_city = economic_df.loc[
    economic_df["Establishments"].idxmax(), "District"
]

# City with highest workforce
top_workforce_city = economic_df.loc[
    economic_df["Workforce"].idxmax(), "District"
]

# City with highest and lowest average item price
cities = ["Islamabad", "Lahore", "Karachi", "Peshawar", "Quetta"]

average_prices = cost_df[cities].mean()

highest_price_city = average_prices.idxmax()
lowest_price_city = average_prices.idxmin()

highest_average_price = average_prices.max()
lowest_average_price = average_prices.min()

# Display final results
print("WEATHER INSIGHTS")
print("Hottest City:", hottest_city)
print("Highest Precipitation:", rainiest_city)

print("\nCOST OF LIVING INSIGHTS")
print("Highest Average Item Price:", highest_price_city, "-", round(highest_average_price, 2))
print("Lowest Average Item Price:", lowest_price_city, "-", round(lowest_average_price, 2))

print("\nECONOMIC INSIGHTS")
print("Most Establishments:", top_establishment_city)
print("Highest Workforce:", top_workforce_city)
```
