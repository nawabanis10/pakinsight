```python
import pandas as pd
from pathlib import Path
```

```python
file_path = "../data/raw/Annex_22.10.2026.xlsx"

xls = pd.ExcelFile(file_path)

print(xls.sheet_names)
```

```python
df = pd.read_excel(
    file_path,
    sheet_name="Appendix-A",
    header=None
)

pd.set_option("display.max_columns", None)

print(df.shape)
```

```python
block_1 = df.iloc[6:57, [1, 2, 4, 16]].copy()

block_2 = df.iloc[64:115, [10, 22]].copy()

block_3 = df.iloc[122:173, [7]].copy()
```

```python
block_1 = block_1.reset_index(drop=True)
block_2 = block_2.reset_index(drop=True)
block_3 = block_3.reset_index(drop=True)
```

```python
combined_data = pd.concat(
    [block_1, block_2, block_3],
    axis=1
)

combined_data.head()
```

```python
combined_data.rename(columns={
    1: "Item",
    2: "Unit",
    4: "Islamabad",
    16: "Lahore",
    10: "Karachi",
    22: "Peshawar",
    7: "Quetta"
}, inplace=True)

combined_data.head()
```

```python
combined_data = combined_data[
    ["Item", "Unit", "Islamabad", "Lahore",
     "Karachi", "Peshawar", "Quetta"]
]

combined_data.head()
```

```python
output_path = Path("../data/Raw")
output_path.mkdir(parents=True, exist_ok=True)

combined_data.to_csv(
    output_path / "pakistan_city_cost_of_living_raw.csv",
    index=False
)
```
