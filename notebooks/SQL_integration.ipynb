```python
from sqlalchemy import create_engine, text
from getpass import getpass
from urllib.parse import quote_plus

password = getpass("Enter MySQL password: ")

engine = create_engine(
    f"mysql+pymysql://root:{quote_plus(password)}@localhost:3306/pakinsight"
)

with engine.connect() as conn:
    result = conn.execute(text("SELECT DATABASE()"))
    print("Connected to:", result.scalar())
```

```python
with engine.connect() as conn:
    result = conn.execute(text("SHOW TABLES"))
    print(result.fetchall())
```

```python
import pandas as pd

df = pd.read_csv("../data/processed/pakistan_weather_2025.csv")

df.head()
```

```python
df.to_sql(
    name="Weather_data",
    con=engine,
    if_exists="fail",
    index=False)
```

```python
with engine.connect() as conn:
    result = conn.execute(text("SELECT COUNT(*) FROM weather_data"))
    print("Total rows:", result.scalar())
```
