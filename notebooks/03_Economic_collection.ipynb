```python
folder = "../data/processed"

lahore_total = pd.read_csv(
    f"{folder}/lahore_economic_total.csv"
)

karachi_east_total = pd.read_csv(
    f"{folder}/karachi_east_economic_total.csv"
)

karachi_west_total = pd.read_csv(
    f"{folder}/karachi_west_economic_total.csv"
)

karachi_central_total = pd.read_csv(
    f"{folder}/karachi_central_economic_total.csv"
)

karachi_south_total = pd.read_csv(
    f"{folder}/karachi_south_economic_total.csv"
)

peshawar_total = pd.read_csv(
    f"{folder}/peshawar_economic_total.csv"
)

quetta_total = pd.read_csv(
    f"{folder}/quetta_economic_total.csv"
)

islamabad_total = pd.read_csv(
    f"{folder}/islamabad_economic_total.csv"
)
```

```python
karachi_districts = pd.concat(
    [
        karachi_east_total,
        karachi_west_total,
        karachi_central_total,
        karachi_south_total
    ],
    ignore_index=True
)

karachi_total = pd.DataFrame({
    "District": ["Karachi (4 districts)"],
    "Establishments": [
        karachi_districts["Establishments"].sum()
    ],
    "Workforce": [
        karachi_districts["Workforce"].sum()
    ]
})

display(karachi_total)
```

```python
economic_summary = pd.concat(
    [
        lahore_total,
        karachi_total,
        peshawar_total,
        quetta_total,
        islamabad_total
    ],
    ignore_index=True
)

display(economic_summary)
```

```python
economic_summary.to_csv(
    output_path / "pakistan_city_economic_summary.csv",
    index=False
)
```

```python
lahore = pd.read_csv(
    f"{folder}/lahore_economic_categories.csv"
)

karachi_east = pd.read_csv(
    f"{folder}/karachi_east_economic_categories.csv"
)

karachi_west = pd.read_csv(
    f"{folder}/karachi_west_economic_categories.csv"
)

karachi_central = pd.read_csv(
    f"{folder}/karachi_central_economic_categories.csv"
)

karachi_south = pd.read_csv(
    f"{folder}/karachi_south_economic_categories.csv"
)

peshawar = pd.read_csv(
    f"{folder}/peshawar_economic_categories.csv"
)

quetta = pd.read_csv(
    f"{folder}/quetta_economic_categories.csv"
)

islamabad = pd.read_csv(
    f"{folder}/islamabad_economic_categories.csv"
)
```

```python
economic_categories = pd.concat(
    [
        lahore,
        karachi_east,
        karachi_west,
        karachi_central,
        karachi_south,
        peshawar,
        quetta,
        islamabad
    ],
    ignore_index=True
)

economic_categories.head()
```

```python
economic_categories["City"] = economic_categories["District"].replace({
    "Karachi East": "Karachi",
    "Karachi West": "Karachi",
    "Karachi Central": "Karachi",
    "Karachi South": "Karachi"
})

economic_categories.head()
```

```python
economic_categories.to_csv(
    output_path / "pakistan_economic_categories.csv",
    index=False
)
```
